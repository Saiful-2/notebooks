## Session 02 — Clean, Validate and Save
### 6. Remove identical copies
Keep one instance of every exact full-row duplicate. The source `df_raw` stays unchanged.

```python
# Remove only exact duplicate rows
df = df_raw.drop_duplicates().copy()
print("Rows after duplicate removal:", len(df))
```

### 7. Standardize text
`strip()` removes extra spaces. `title()` makes text capitalization consistent; we then correct known spelling variants. Do **not** guess missing crop or pest answers.

```python
# Standardize text columns (except IDs)
text_cols = ["division", "district", "season", "crop", "seed_type",
             "soil_type", "irrigation_method", "pest_observed"]
for col in text_cols:
    df[col] = df[col].astype("string").str.strip().str.title()

# Preserve ID format; do not treat repeated farm IDs as duplicates
df["farm_id"] = df["farm_id"].astype("string").str.strip().str.upper()

# Treat common placeholders as missing
df[text_cols] = df[text_cols].replace(
    ["Unknown", "Not Recorded", "?", "-"], pd.NA)
```

```python
# Correct known aliases from this teaching dataset
df["crop"] = df["crop"].replace(
    {"Rcie": "Rice", "Mzie": "Maize", "Pototo": "Potato", "Whaet": "Wheat", "Jtue": "Jute"})
df["season"] = df["season"].replace(
    {"Amman": "Aman", "Borro": "Boro", "Aush": "Aus", "Rabbi": "Rabi", "Kharif 1": "Kharif-I"})
df["soil_type"] = df["soil_type"].replace(
    {"Loamy": "Loam", "Clayey": "Clay", "Silty Loam": "Silt loam",
     "Silt Loam": "Silt loam", "Sandy-Loam": "Sandy loam", "Sandy Loam": "Sandy loam"})
df["district"] = df["district"].str.rstrip("?").replace(
    {"Barisal": "Barishal", "Bogra": "Bogura", "Cumila": "Cumilla",
     "Dhkaa": "Dhaka", "Jessore": "Jashore", "Jhenaida": "Jhenaidah", "Mymensing": "Mymensingh",
     "Cox'S Bazar": "Cox's Bazar"})
df["pest_observed"] = df["pest_observed"].replace({"Y": "Yes", "N": "No"})
# "Moderate" or "Severe" is not a Yes/No answer; keep it missing
df["pest_observed"] = df["pest_observed"].where(
    df["pest_observed"].isin(["Yes", "No"]))
```

### 8. Convert numbers and dates
Some numerical cells contain text such as `0.517 ha` or `2,400 mm`. We remove these *known teaching-data units* and use `to_numeric()`. For other real datasets, always check units before converting.

`to_datetime()` converts mixed date formats; invalid dates become missing (`NaT`).

```python
# Convert the eight measurement columns to numbers
number_cols = ["soil_ph", "annual_rainfall_mm", "annual_mean_temp_c",
               "irrigation_mm", "fertilizer_kg_ha", "area_ha",
               "production_ton", "yield_ton_ha"]
for col in number_cols:
    numbers = df[col].astype("string").str.replace(r"[^0-9.\-]", "", regex=True)
    df[col] = pd.to_numeric(numbers, errors="coerce")

print(df[number_cols].dtypes)
```

```python
# Read dates; errors='coerce' marks impossible dates as missing
for col in ["planting_date", "harvest_date"]:
    df[col] = pd.to_datetime(df[col], format="mixed", dayfirst=True, errors="coerce")

print(df[["planting_date", "harvest_date"]].head())
```

### 9. Handle missing values responsibly
`fillna()` can display missing pest answers as **Unknown**, without inventing `Yes` or `No`. `dropna()` *could* remove incomplete records, but we will only separate records with important unresolved problems later.

```python
# Display unanswered pest observations without changing the source
a = df["pest_observed"].fillna("Unknown")
print(a.value_counts())

# Example: number of records with a known soil pH
print("Known soil pH:", len(df.dropna(subset=["soil_ph"])))
```

### 10. Check realistic numerical ranges
These are **broad classroom screening ranges**, not official crop thresholds. We **flag** suspect values instead of automatically replacing them. A true extreme can be valid.

```python
# Flag non-missing measurements outside broad screening ranges
limits = {"soil_ph": (3.5, 9.5), "annual_rainfall_mm": (500, 6500),
          "annual_mean_temp_c": (15, 35), "irrigation_mm": (0, 3000),
          "fertilizer_kg_ha": (0, 2000), "area_ha": (0.01, 20),
          "production_ton": (0.001, 1000), "yield_ton_ha": (0.1, 50)}
bad_range = pd.Series(False, index=df.index)
for col, (low, high) in limits.items():
    bad_range |= df[col].notna() & ~df[col].between(low, high)
print("Out-of-range records:", bad_range.sum())
```

### 11. Outliers: investigate, do not automatically delete
The **IQR rule** highlights unusual observations. Compare a *single crop and season* (e.g., Boro rice), because potato and rice have very different yield scales.

```python
# IQR check for Boro rice yield only
boro = df.loc[(df["crop"] == "Rice") & (df["season"] == "Boro"), "yield_ton_ha"].dropna()
q1, q3 = boro.quantile([0.25, 0.75])
iqr = q3 - q1
outliers = (boro < q1 - 1.5 * iqr) | (boro > q3 + 1.5 * iqr)
print("Boro yield values to investigate:", outliers.sum())
```

### 12. Validate agricultural logic
The following flags find inconsistencies that we cannot safely correct by guessing. **Boro can be planted in the previous calendar year**, so compare `harvest_year` to the *harvest*, not the planting year.

```python
# Dates must be complete, ordered and agree with harvest year
bad_dates = (df["planting_date"].isna() | df["harvest_date"].isna() |
             (df["harvest_date"] <= df["planting_date"]) |
             (df["harvest_date"].dt.year != df["harvest_year"]))

# Check permitted crop–season combinations
allowed = ["Rice-Boro", "Rice-Aman", "Rice-Aus", "Wheat-Rabi",
           "Maize-Rabi", "Maize-Kharif-I", "Potato-Rabi", "Jute-Kharif-I"]
bad_season = ~(df["crop"] + "-" + df["season"]).isin(allowed)
print("Date issues:", bad_dates.sum(), " | Crop–season issues:", bad_season.sum())
```

```python
# Output ≈ area × yield (allow for decimal rounding)
bad_output = (df["production_ton"] / df["area_ha"] - df["yield_ton_ha"]).abs() > 0.05

# Rainfed means no applied irrigation; other methods should have a positive depth
bad_irrigation = ((df["irrigation_method"] == "Rainfed") & df["irrigation_mm"].gt(0)) | (
    df["irrigation_method"].isin(["Pump", "Canal"]) & df["irrigation_mm"].le(0))

# IDs must not conflict; farm ID format = FM-DD-NNN
bad_id = df["record_id"].duplicated(keep=False) | ~df["farm_id"].str.fullmatch(
    r"FM-\d{2}-\d{3}").fillna(False)
print("Output:", bad_output.sum(), " | Irrigation:", bad_irrigation.sum(), " | IDs:", bad_id.sum())
```

```python
# Each district should belong to one division
usual_division = df.groupby("district")["division"].transform(lambda x: x.mode().iloc[0])
bad_location = df["division"].ne(usual_division).fillna(True)

# Annual rainfall should repeat within each district + year (not be summed)
usual_rain = df.groupby(["district", "harvest_year"])["annual_rainfall_mm"].transform("median")
bad_weather = (df["annual_rainfall_mm"] - usual_rain).abs() > 1
print("Location:", bad_location.sum(), " | Weather conflicts:", bad_weather.sum())
```

### 13. Separate unresolved records
We can correct standardized text and data formats confidently. We **do not guess** unknown dates, production values, or irrigation totals. Records with major unresolved issues go into a separate review CSV; records with noncritical missing fields (e.g., soil pH or pest status) may remain, clearly marked as missing.

**Important:** This is a **beginner-level screening workflow**, not a certified reconstruction of every invented error. The instructor's private reference is not used in this student notebook. Some problems can remain undetected.

```python
# Core fields needed for reliable crop and yield analysis
required = ["record_id", "farm_id", "division", "district", "crop", "season",
            "planting_date", "harvest_date", "area_ha", "production_ton", "yield_ton_ha",
            "irrigation_method"]
missing_core = df[required].isna().any(axis=1)

# Send uncertain observations for review; retain the original 21 columns
needs_review = (missing_core | bad_range | bad_dates | bad_season |
                bad_output | bad_irrigation | bad_id | bad_location | bad_weather).fillna(True)
df_clean = df.loc[~needs_review].copy()
df_review = df.loc[needs_review].copy()
print("For analysis:", len(df_clean), " | Needs review:", len(df_review))
```
